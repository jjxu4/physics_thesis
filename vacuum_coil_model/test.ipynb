```python
from current_loop import current_loop
from loop_system import loop_system
import numpy as np
from IPython.display import HTML

def current(t):
    return np.sin(t)

def dIdt(t):
    return np.cos(t)

loop = current_loop("loop", 10e-2, 0, current, dIdt, 2e-3)

times = np.linspace(0, 20, 2)
R = np.linspace(0, 1, 4)
Z = np.linspace(-0.5, 0.5, 3)
R_grid, Z_grid = np.meshgrid(R, Z)

print(R_grid)
print(Z_grid)
print(times)

my_loops = loop_system([loop])
E_field_total = my_loops.total_Efield_phi(times, R_grid, Z_grid)
print(E_field_total)

ani = my_loops.animate_loop_voltage(0.333, 0, 20, 2, 2, -0.5, 0.5, 3, 1, 4)
HTML(ani.to_jshtml())
```
